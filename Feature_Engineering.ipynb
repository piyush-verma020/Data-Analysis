# Feature Engineering Process

Step 1: `Exploratory Data Analysis (EDA)`

After collecting raw data, we first explore it to understand its characteristics:

* Numerical Features → visualized with histograms, probability density functions, etc.

* Categorical Features → checked for distribution and unique categories.

* Missing Values → identified and visualized (heatmaps, bar plots).

* Outliers → detected using boxplots and statistical methods.

* Data Cleaning → deciding whether to drop, impute, or transform noisy data.

* This step helps us convert raw data into useful data that improves model performance.


Step 2: `Handling Missing Values`

* Mean, Median, Mode imputation

* Advanced methods: KNN imputer, regression imputation, etc.

Step 3: `Handling Imbalanced Data`

* Oversampling (SMOTE)

* Undersampling

* Class weights

Step 4: `Treating Outliers`

* Removing, capping, or transforming (log, Box-Cox) outliers.

Step 5: `Feature Scaling`

* Standardization (z-score normalization)

* Min-Max Normalization

Step 6: `Encoding Categorical Features`

* One-Hot Encoding

* Label Encoding

* Target Encoding (depending on algorithm).

### Importance of Feature Engineering:

`Raw Data` → `Feature Engineering` → `Clean Data` → `Machine Learning Model` → `Better Predictions`

# Feature Selection

After feature engineering, we reduce dimensionality and keep only the most useful features.

##### Techniques for Feature Selection:

* Correlation Analysis

* Chi-Square Test

* KNN-based methods

* Genetic Algorithms

* Feature Importance (using ExtraTrees Classifier, Random Forest, XGBoost, etc.)
